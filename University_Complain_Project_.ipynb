```python

import json

def load_complaints():

    try:
        with open("complaints.json", "r") as file:
            complaints = json.load(file)

    except FileNotFoundError:
        complaints = []

    except json.JSONDecodeError:
        complaints = []

    return complaints

def save_complaints(complaints):

    with open("complaints.json", "w") as file:
        json.dump(complaints, file, indent=4)

def submit_complaint():

    complaints = load_complaints()

    print("\n--- Submit Complaint ---")

    name = input("Enter your name: ")
    student_id = input("Enter Student ID: ")
    department = input("Enter Department: ")
    complaint = input("Enter your complaint: ")

    new_complaint = {
        "Name": name,
        "Student ID": student_id,
        "Department": department,
        "Complaint": complaint,
        "Status": "Pending"
    }

    complaints.append(new_complaint)

    save_complaints(complaints)

    print("\nYour complaint has been submitted successfully!")


def view_complaints():

    complaints = load_complaints()

    student_id = input("Enter your Student ID: ")

    found = False

    for c in complaints:

        if c["Student ID"] == student_id:

            print("\n--- Your Complaint ---")
            print("Name:", c["Name"])
            print("Student ID:", c["Student ID"])
            print("Department:", c["Department"])
            print("Complaint:", c["Complaint"])
            print("Status:", c["Status"])

            found = True

    if found == False:
        print("\nNo complaint found!")

def view_all_complaints():

    complaints = load_complaints()

    if len(complaints) == 0:

        print("\nNo complaints available.")

    else:

        for c in complaints:

            print("\n--- Complaint Details ---")
            print("Name:", c["Name"])
            print("Student ID:", c["Student ID"])
            print("Department:", c["Department"])
            print("Complaint:", c["Complaint"])
            print("Status:", c["Status"])


def search_complaint():

    complaints = load_complaints()

    student_id = input("Enter Student ID to search: ")

    found = False

    for c in complaints:

        if c["Student ID"] == student_id:

            print("\n--- Complaint Found ---")
            print("Name:", c["Name"])
            print("Student ID:", c["Student ID"])
            print("Department:", c["Department"])
            print("Complaint:", c["Complaint"])
            print("Status:", c["Status"])

            found = True

    if found == False:

        print("\nComplaint not found!")



def admin_login():

    password = input("Enter Admin Password: ")

    if password == "1234":

        print("\nAdmin Login Successful!")

        while True:

            print("\n===== ADMIN MENU =====")
            print("1. View All Complaints")
            print("2. Search Complaint")
            print("3. Exit Admin Menu")

            choice = input("Enter your choice: ")

            if choice == "1":

                view_all_complaints()

            elif choice == "2":

                search_complaint()

            elif choice == "3":

                print("\nLeaving Admin Menu...")
                break

            else:

                print("\nInvalid choice! Try again.")

    else:

        print("\nIncorrect Password!")



def main():

    while True:

        print("\n===== UNIVERSITY COMPLAINT PORTAL =====")
        print("1. Submit Complaint")
        print("2. View My Complaints")
        print("3. Admin Login")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":

            submit_complaint()

        elif choice == "2":

            view_complaints()

        elif choice == "3":

            admin_login()

        elif choice == "4":

            print("\nThank you for using the portal!")
            break

        else:

            print("\nInvalid choice! Try again.")


main()
```
